# Ingestão dos Indicadores INEP

## 1. Configuração

```python
import polars as pl
from pathlib import Path

# Caminhos principais do projeto
ROOT = Path.cwd().parent if Path.cwd().name == "src" else Path.cwd()
RAW = ROOT / "data" / "raw"
INDICADORES = ROOT / "data" / "processed" / "indicadores"
ENEM = ROOT / "data" / "processed" / "enem"

INDICADORES.mkdir(parents=True, exist_ok=True)
```

## 2. Indicadores do Censo Escolar 2024

```python
# AFD: adequação docente | ATU: alunos por turma
# DSU: docentes com superior | TDI: distorção idade-série
for sigla in ["afd", "atu", "dsu", "tdi"]:
    origem = RAW / f"raw_{sigla}_2024.xlsx"
    destino = INDICADORES / f"data_{sigla}.parquet"
    pl.read_excel(origem, engine="calamine").write_parquet(destino)
    print(f"{origem.name} -> {destino.name}")
```

## 3. INSE 2023 e Adequação por Disciplina/UF 2024

```python
# INSE 2023 por escola (aba INSE_ESC_2023)
pl.read_excel(
    RAW / "INSE_2023_escolas.xlsx", sheet_name="INSE_ESC_2023", engine="calamine"
).write_parquet(INDICADORES / "data_inse.parquet")
print("INSE_2023_escolas.xlsx -> data_inse.parquet")

# Taxa de inadequação docente em Física, Química e Biologia por UF
pl.read_excel(
    RAW / "Adequacao2024.xlsx", engine="calamine"
).write_parquet(INDICADORES / "data_adequacao_uf.parquet")
print("Adequacao2024.xlsx -> data_adequacao_uf.parquet")
```
