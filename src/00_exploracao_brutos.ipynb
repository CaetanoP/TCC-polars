# Exploração dos Dados Brutos

## 1. Configuração

```python
import polars as pl
from pathlib import Path

# Caminhos principais do projeto
ROOT = Path.cwd().parent if Path.cwd().name == "src" else Path.cwd()
RAW = ROOT / "data" / "raw"
INDICADORES = ROOT / "data" / "processed" / "indicadores"
ENEM = ROOT / "data" / "processed" / "enem"

pl.Config.set_tbl_cols(15)
```

## 2. Planilhas Complementares (.xlsx)

### 2.1 INSE 2023 e Adequação por UF 2024

```python
# INSE por escola e taxa de inadequação docente por disciplina/UF
df_inse = pl.read_excel(RAW / "INSE_2023_escolas.xlsx", sheet_name="INSE_ESC_2023", engine="calamine")
df_adeq = pl.read_excel(RAW / "Adequacao2024.xlsx", engine="calamine")

print("INSE_2023_escolas.xlsx:", df_inse.shape)
display(df_inse.head(3))
print("Adequacao2024.xlsx:", df_adeq.shape)
display(df_adeq.head(5))
```

### 2.2 Provas de Ciências da Natureza (.xlsx)

```python
# Gabaritos por cor de prova com marcação de Física, Química e Biologia
for arquivo in sorted((RAW / "Provas").glob("*.xlsx")):
    df = pl.read_excel(arquivo, engine="calamine")
    print(arquivo.name, df.shape)
    display(df.head(3))
```

## 3. Microdados ENEM 2024 (.csv)

```python
# Leitura lazy das primeiras linhas (arquivos de até 1.6 GB)
for arquivo in sorted((RAW / "microdados_enem_2024").glob("*.csv")):
    df = pl.scan_csv(arquivo, separator=";", encoding="utf8-lossy").head(5).collect()
    print(arquivo.name, f"{df.width} colunas")
    display(df)
```
